### Understanding Database Relationships (1:1 and 1:N)
In our PACE Skaters Hub, relationships define how our tables "talk" to each other. We use two main types:

1. One-to-One (1:1)
Definition: Each record in Table A relates to exactly one record in Table B, and vice versa.

The Example: Member ↔ Student ID Card

The Logic: One member can have only one official student ID card, and that specific ID card belongs to only one person.

In the Diagram: You usually use this when splitting a table into two for security or organization (e.g., keeping sensitive home addresses in a separate table).

2. One-to-Many (1:N)
Definition: A single record in Table A (the "Parent") can relate to multiple records in Table B (the "Children"), but each "Child" has only one "Parent."

The Example: Club Member ↔ Skating Logs

The Logic: * One Parent: One member (e.g., Cesar) can go skating many times.

Many Children: Each session recorded in the member_rink_log belongs to only one specific person.

The Example: Rinks ↔ Skating Logs

One rink (e.g., The Forks) can host hundreds of skating sessions. However, one specific session happens at only one rink at a time.

#### DDL vs. DML: The Developer's Vocabulary

Now, we need to create these five tables in PostgreSQL. In the world of databases, we categorize our actions into two main "fancy" terms:

1.  **DDL (Data Definition Language)**: This refers to **creating** or **modifying** the table structure (e.g., creating a table or adding a new column).
2.  **DML (Data Manipulation Language)**: This refers to handling the actual data inside the tables: **C**reate (Insert), **R**ead (Select), **U**pdate, and **D**elete—commonly known as **CRUD**.

> **A Reality Check:** You’ll often hear senior developers joke that even after 10 years, their daily job is still mostly "CRUD." Don't feel discouraged! It's perfectly normal because managing data correctly is the most critical part of any system.

```python
%%sql
-- 1. Create a member table (Parent Table)
CREATE TABLE club_member (
id SERIAL PRIMARY KEY,
name VARCHAR(100) NOT NULL,
phone_number VARCHAR(20),
email_address VARCHAR(100) UNIQUE,
age INT CHECK (age >= 0),
enroll_date DATE DEFAULT CURRENT_DATE
);

-- 2. Create ice rink table (Parent Table)
CREATE TABLE rinks_place (
id SERIAL PRIMARY KEY,
name VARCHAR(100) NOT NULL,
address varchar(2048),
postal_code VARCHAR(10),
rental_valid BOOLEAN DEFAULT TRUE
);

CREATE TABLE skills (
id SERIAL PRIMARY KEY,
skill_name VARCHAR(100) NOT NULL -- like Duck Walk
);

CREATE TABLE skill_level (
id SERIAL PRIMARY KEY,
skill_id INT REFERENCES skills(id), -- 1:N relationship
level_number INT UNIQUE NOT NULL, -- such as 1, 2, 3
level_name VARCHAR(50) -- such as Beginner, Intermediate
);

-- 5. Create a skating record sheet (The "Heart" - Child of Member, Rink, and Level)
CREATE TABLE member_rink_log (
id SERIAL PRIMARY KEY,
member_id INT REFERENCES club_member(id) ON DELETE CASCADE,
position INT REFERENCES rinks_place(id),
level_id INT REFERENCES skill_level(id),
activity_date TIMESTAMP WITH TIME ZONE DEFAULT CURRENT_TIMESTAMP,
commute_time INTERVAL, -- PG-specific time interval type, records how long the slide lasted
expense NUMERIC(10, 2) -- Records fuel costs or entrance fees, more accurate than Float

);
```

The Reality of Big Data: Why Pagination Matters
Years later, after participating in hundreds of events, our database has grown from a few rows to tens of thousands of records. In a real industrial environment, it’s common for a single table to exceed millions of rows.

At this scale, running a simple SELECT * FROM member_rink_log is no longer just a query—it’s a performance disaster. Returning millions of rows in one go would:

Choke the Network: The massive Data Payload would slow down the transfer.

Crash the App: The application memory might overflow trying to process all that data.

Kill the UX: No user can (or wants to) scroll through a million rows of data.

This is where Pagination Queries become a mandatory standard in backend engineering.
